```python
# Cleans numbers string with decimal sesperateors, turns into usable integer
def clean_data(column_data):
    return column_data.str.replace(',','').astype(int)
```

```python

clean_num_of_businesses = clean_data(df["Number of Businesses"])
```

```python
mean_NOB = clean_num_of_businesses.mean()
median_NOB = clean_num_of_businesses.median()
mode_NOB = clean_num_of_businesses.mode()
range_NOB = clean_num_of_businesses.max() - clean_num_of_businesses.min() # gets range

# variance
var_NOB = clean_num_of_businesses.var() 

# standard deviation
std_NOB = clean_num_of_businesses.std() 

# calculates both 
q1_NOB, q3_NOB = clean_num_of_businesses.quantile([0.25, 0.75]) 

# interquartile range
IQR_NOB = q3_NOB - q1_NOB 

# nice stats structure for all number of businesses statistics
# we dont round in case we need to calculate something with the values later
NOB_stats = {
    "mean": mean_NOB,
    "median": median_NOB,
    "mode(s)": mode_NOB,
    "range": range_NOB,
    "std": std_NOB,
    "variance": var_NOB,
    "quartile 1": q1_NOB,
    "quartile 3": q3_NOB,
    "IQR": IQR_NOB,
    
}
```

```python
# Shows our stats for Number of Businesses"!
print("Stats for Number of Businesses")
for key, value in NOB_stats.items():
    print(key + ": " + str(value))
```

```python
# number I chose to calculate z-score with this specific data point, sdp for short
# Its more useful if the data is sorted
sdp_NOB = sorted(clean_num_of_businesses)[300]
```

```python
# to get the z-score , we calculate using z = (value - mean)/std
z_score_sdp_NOB = (sdp_NOB - NOB_stats["mean"])/NOB_stats["std"]
z_score_sdp_NOB
# thats pretty close to the mean, only 0.29 standard deviations awa from the mean
```

### REFLECTION ###
The Median best describes the Number Of Employees.
Since the mean is dramatically higher than the median, it gives a false impression that neighborhood businesses collectively have more than 1000 employees. In reality, they have around half that much.

I learned that the first quartile of the number of businesses is very low compared with the third quartile, and quite close to 0. This indicates a notable number of neighborhoods with only a few busineses. We might interpret those areas to be residential neighborhoods with a handfull of corner stores or bodegas.
Since we know the mean is larger than the median, we can infer there some outlier neighborhoods with a large number of businesses. We could assume these areas might have malls with many small businesses inside.

For the Number of Employees, the large number of upper bound outliers give the impression of a class of neighborhood dedicated to large workforces, such as a downtown area with many office buildings, or an industrial park.

I must note the caveat that there is a Survey Year value, which could possibly imply that neighborhoods appear each year, or only appear some years. Without being able to examine neightborhoods, businesses or employees by year, we cannot assume too much with confidence.
